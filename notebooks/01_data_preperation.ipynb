## Data Dictionary — Global Power Plant Database (WRI, v1.3.0)

**Source:** World Resources Institute (WRI) — Global Power Plant Database  
**File:** `data/raw/global_power_plant_database.csv`  
**Size:** 34,936 rows (power plants) × 36 columns

### Identification & location

| # | Column | Type | Meaning |
|---|---|---|---|
| 1 | `country` | text | 3-letter country code (ISO 3166-1 alpha-3), e.g. `USA`, `SAU` |
| 2 | `country_long` | text | Full country name |
| 3 | `name` | text | Name of the power plant |
| 4 | `gppd_idnr` | text | Unique ID of the plant in the database |
| 5 | `latitude` | float | Latitude of the plant (decimal degrees, WGS84) |
| 6 | `longitude` | float | Longitude of the plant (decimal degrees, WGS84) |
| 7 | `geolocation_source` | text | Organization or method that provided the coordinates |

### Capacity & fuel

| # | Column | Type | Meaning |
|---|---|---|---|
| 8 | `capacity_mw` | float | Electrical generating capacity in megawatts (MW): the maximum power the plant can produce |
| 9 | `year_of_capacity_data` | float | Year in which the `capacity_mw` value was reported |
| 10 | `primary_fuel` | text | Main energy source: Coal, Gas, Oil, Hydro, Solar, Wind, Nuclear, Biomass, Geothermal, Waste, Storage, Cogeneration, Petcoke, Wave and Tidal, Other |
| 11 | `other_fuel1` | text | First additional fuel the plant can use (if any) |
| 12 | `other_fuel2` | text | Second additional fuel (if any) |
| 13 | `other_fuel3` | text | Third additional fuel (if any) |
| 14 | `commissioning_year` | float | Year the plant started operating. Can be a decimal when units started in different years (capacity-weighted average) |

### Ownership & sources

| # | Column | Type | Meaning |
|---|---|---|---|
| 15 | `owner` | text | Majority owner of the plant |
| 16 | `source` | text | Organization that reported the plant's data |
| 17 | `url` | text | Web link to the data source |
| 18 | `wepp_id` | text | Matching ID in the Platts WEPP database (a commercial dataset), used for cross-referencing |

### Reported generation (real data, GWh)

| 25 | `generation_gwh_2019` | float | Reported electricity generated in 2019 (GWh) |

### Estimated generation (WRI model, GWh)

| # | Column | Type | Meaning |
|---|---|---|---|
| 27 | `estimated_generation_gwh_2013` | float | Model-estimated electricity generated in 2013 (GWh) |
| 28 | `estimated_generation_gwh_2014` | float | Model-estimated electricity generated in 2014 (GWh) |
| 29 | `estimated_generation_gwh_2015` | float | Model-estimated electricity generated in 2015 (GWh) |
| 30 | `estimated_generation_gwh_2016` | float | Model-estimated electricity generated in 2016 (GWh) |
| 31 | `estimated_generation_gwh_2017` | float | Model-estimated electricity generated in 2017 (GWh) |
| 32 | `estimated_generation_note_2013` | text | Estimation method used for 2013 (see codes below) |
| 33 | `estimated_generation_note_2014` | text | Estimation method used for 2014 |
| 34 | `estimated_generation_note_2015` | text | Estimation method used for 2015 |
| 35 | `estimated_generation_note_2016` | text | Estimation method used for 2016 |
| 36 | `estimated_generation_note_2017` | text | Estimation method used for 2017 |

### Estimation method codes (`estimated_generation_note_*`)

| Code | Meaning |
|---|---|
| `CAPACITY-FACTOR-V1` | Generic method: capacity × typical capacity factor for that fuel/country × 8,760 hours |
| `SOLAR-V1` | Solar model based on location/sunlight, including plant age |
| `SOLAR-V1-NO-AGE` | Same solar model, without plant age (commissioning year missing) |
| `WIND-V1` | Wind model based on location/wind speed |
| `HYDRO-V1` | Hydro model based on location/water flow |
| `NO-ESTIMATION` | No estimate produced, so the estimated value is empty |

### Notes
- **Units:** 1 GWh = 1,000 MWh. Maximum yearly output ≈ `capacity_mw × 8,760 / 1,000` GWh.
- **Reported vs estimated:** `generation_gwh_*` is real reported data, available for only ~9,500 plants. `estimated_generation_gwh_*` is modelled and available for most plants (2013–2017 only).
- **Mostly empty columns:** `other_fuel2`, `other_fuel3`, `wepp_id`, `owner` and the generation columns are largely empty. Check with `df.isna().mean()`.
